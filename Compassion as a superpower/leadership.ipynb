# Be a leader. Don't try to lead

- Leading up and down the chain of command (Extreme Ownership, chapter 10)
- Leadership is omnidirectional
- "I don't hire smart people to tell them what to do. I hire smart people to tell me what to do."- Steve Jobs
- You will understand situations better than your boss sometimes. Lead them to a solution. Educate them. 
  - Works only with a good "boss"
  - **or** great skill on your part
- Easier is to allow yourself to be led. Do that.

# Other lessons

# Only apologize if you are authentic (The Next Conversation)

- Instead, thank the other person for their accommodation. 
- Apologize when you've caused harm. Then, fix it. Be sincere. Some things cannot be fixed. You will have to live with them. 
- Your guilt is the price you pay for your growth.

# Take ownership

- This does not mean to do everything
  - It means that when you are assigned a task, you ensure that it does get done, by you or someone else.
- Everything your team must accomplish, you own.

# Every failure means:
  - **You** assigned the wrong person
  - **You** did not train them well enough
  - **You** didn't clarify the assignment
  - **You** didn't provide the resources
  - **You** didn't provide the time

# Provide backup (variation on Extreme Ownership- chapter 5)

- Never assign a project without
  - A backup person
  - A schedule that allows time to bring in one or the other
  - Assessment of progress in time to do so
- For clubs, always assign a senior member and a junior member to the task.
  - If it goes poorly, you have a backup, and a massive growth opportunity (be ready to help the junior member)
